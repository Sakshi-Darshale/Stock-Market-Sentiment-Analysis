**Objectives:**

5. Save models and vectorizers to `models/`.


```
import sys
from pathlib import Path
import pandas as pd
import numpy as np
import joblib

sys.path.insert(0, str(Path("../").resolve()))
from src.train_model import train_models, prepare_data
```

## 1. Execute Supervised Training Pipeline

```
# Train models and save serialized artifacts to models/
results = train_models(save_models=True)
summary_table = results["summary"]
display(summary_table)
```

## 2. Inspect TF-IDF Feature Space

```
vectorizer = results["vectorizer"]
feature_names = vectorizer.get_feature_names_out()

print(f"Total Unique Vocabulary Features: {len(feature_names)}")
print("Sample N-Gram Features:", list(feature_names[:25]))
```

## 3. Verify Serialized Model Artifacts

```
models_dir = Path("../models")
for artifact in models_dir.glob("*.pkl"):
    size_kb = artifact.stat().st_size / 1024
    print(f"Artifact: {artifact.name:30s} | Size: {size_kb:.2f} KB")
```
