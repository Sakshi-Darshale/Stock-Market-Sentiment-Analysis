# Notebook 01: Exploratory Data Analysis (EDA)
### Project: Stock Market Sentiment Analysis (TY BSc Data Science)

**Objectives:**
1. Ingest and validate raw financial news dataset (`data/raw/financial_news.csv`) and stock price quotes (`data/raw/stock_data.csv`).
2. Analyze class distributions across Positive, Negative, and Neutral sentiment.
3. Examine coverage by enterprise ticker (AAPL, MSFT, GOOGL, AMZN, TSLA, NVDA).
4. Inspect historical stock price movements and volume trends.


```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

# Setup plotting style
plt.style.use('seaborn-v0_8-whitegrid' if 'seaborn-v0_8-whitegrid' in plt.style.available else 'default')
%matplotlib inline
```

## 1. Load Raw Datasets

```
# Define paths
data_dir = Path("../data/raw")
news_df = pd.read_csv(data_dir / "financial_news.csv")
stock_df = pd.read_csv(data_dir / "stock_data.csv")

print(f"Financial News Records: {len(news_df)}")
print(f"Historical Stock Records: {len(stock_df)}")
news_df.head()
```

## 2. Financial News Sentiment Distribution

```
# Class distribution counts and percentages
sentiment_counts = news_df['sentiment'].value_counts()
sentiment_pcts = news_df['sentiment'].value_counts(normalize=True) * 100

summary_df = pd.DataFrame({'Count': sentiment_counts, 'Percentage (%)': sentiment_pcts.round(2)})
display(summary_df)

# Plot Class Balance
colors = {'Positive': '#00C853', 'Neutral': '#FFAB00', 'Negative': '#D50000'}
plt.figure(figsize=(7, 4))
sns.barplot(x=sentiment_counts.index, y=sentiment_counts.values, palette=colors)
plt.title("Financial Headline Sentiment Distribution", fontsize=13, fontweight='bold')
plt.xlabel("Sentiment Class", fontsize=11)
plt.ylabel("Number of Headlines", fontsize=11)
plt.show()
```

## 3. Headline Coverage by Enterprise Ticker

```
plt.figure(figsize=(9, 4))
sns.countplot(data=news_df, x='ticker', hue='sentiment', palette=colors)
plt.title("Sentiment Mentions by Company Ticker", fontsize=13, fontweight='bold')
plt.xlabel("Ticker", fontsize=11)
plt.ylabel("Headline Count", fontsize=11)
plt.legend(title="Sentiment")
plt.show()
```

## 4. Historical Stock Price Overview

```
stock_df['date'] = pd.to_datetime(stock_df['date'])

plt.figure(figsize=(12, 5))
for ticker in stock_df['ticker'].unique():
    subset = stock_df[stock_df['ticker'] == ticker]
    plt.plot(subset['date'], subset['close'], label=ticker, linewidth=1.8)

plt.title("Historical Stock Closing Prices by Ticker", fontsize=13, fontweight='bold')
plt.xlabel("Date", fontsize=11)
plt.ylabel("Closing Price (USD)", fontsize=11)
plt.legend(title="Ticker", loc='upper left')
plt.show()
```

## 5. Key EDA Findings
- The financial news corpus presents balanced representation across the three sentiment classes.
- Stock price quotes span consistent chronological dates matching the news timestamps.
