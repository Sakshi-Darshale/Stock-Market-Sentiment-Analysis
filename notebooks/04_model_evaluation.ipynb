```
import sys
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sys.path.insert(0, str(Path("../").resolve()))
from src.model_evaluation import evaluate_all_models
from src.explainability import get_global_feature_importance, explain_prediction
```

## 1. Run Complete Evaluation Suite

```
eval_results = evaluate_all_models(save_outputs=True)
comparison_df = eval_results["comparison_table"]
display(comparison_df)
```

## 2. Model Performance Comparison Chart

```
plt.figure(figsize=(9, 4.5))
sns.barplot(data=comparison_df, x='Model', y='Macro F1', palette='viridis')
plt.title("Macro-F1 Score Comparison Across Models", fontsize=13, fontweight='bold')
plt.ylabel("Macro-Averaged F1 Score", fontsize=11)
plt.ylim(0, 1.1)
plt.show()
```

## 3. Explainable AI: Top Global Predictive Features

```
# Extract top feature weights for Logistic Regression
importance_by_class = get_global_feature_importance("Logistic Regression", top_n=8)

fig, axes = plt.subplots(1, 3, figsize=(16, 5))
classes = ["Negative", "Neutral", "Positive"]
colors = ['#D50000', '#FFAB00', '#00C853']

for ax, cls, col in zip(axes, classes, colors):
    df_cls = importance_by_class[cls]
    pos_terms = df_cls[df_cls["direction"] == "Pushes Toward Class"]
    sns.barplot(data=pos_terms, x='weight', y='feature', ax=ax, color=col)
    ax.set_title(f"Top Features Driving: {cls}", fontsize=12, fontweight='bold')
    ax.set_xlabel("Model Coefficient Weight")
    ax.set_ylabel("")

plt.tight_layout()
plt.show()
```

## 4. Local Headline Attribution Test

```
sample_headline = "Apple reports blowout earnings and record operating margin expansion."
xai = explain_prediction(sample_headline, model_name="Logistic Regression")

print(f"Test Headline:     {sample_headline}")
print(f"Predicted Class:   {xai['predicted_class']} (Confidence: {xai['confidence']:.1%})")
print(f"Explanation:       {xai['summary_explanation']}\n")

contrib_df = pd.DataFrame(xai['word_contributions'])
display(contrib_df)
```
