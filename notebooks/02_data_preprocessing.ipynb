# Notebook 02: NLP Data Preprocessing
### Project: Stock Market Sentiment Analysis (TY BSc Data Science)

**Objectives:**
1. Text normalization: Lowercasing, URL removal, HTML removal, punctuation stripping.
2. Tokenization and sentiment-aware stop-word removal (preserving direction and negation words).
3. WordNet lemmatization for lexical standardisation.
4. Export preprocessed dataset to `data/processed/cleaned_news.csv`.


```
import sys
from pathlib import Path
import pandas as pd
import nltk

# Add project src to path
sys.path.insert(0, str(Path("../").resolve()))
from src.data_preprocessing import clean_text, tokenize_and_lemmatize, preprocess_text, load_and_preprocess_dataset
```

## 1. Demonstrate Preprocessing Steps on Raw Financial Text

```
raw_sample = "NVIDIA's ($NVDA) gross margins widen to 78.4% beating Wall Street expectations! https://reuters.com/nvda"
print("1. Raw Input Text:")
print("  ", raw_sample)

cleaned = clean_text(raw_sample)
print("\n2. Regex Cleaned Text (Lowercased, no URLs, no symbols):")
print("  ", cleaned)

tokens = tokenize_and_lemmatize(cleaned)
print("\n3. Lemmatized Tokens (Stop-words removed, negations preserved):")
print("  ", tokens)

final_text = preprocess_text(raw_sample)
print("\n4. Final Unified Preprocessed Text:")
print("  ", final_text)
```

## 2. Batch Preprocessing on Corpus

```
# Run pipeline and export cleaned dataset
processed_df = load_and_preprocess_dataset(
    input_path=Path("../data/raw/financial_news.csv"),
    output_path=Path("../data/processed/cleaned_news.csv")
)

print(f"Total Processed Records: {len(processed_df)}")
processed_df[['text', 'cleaned_text', 'sentiment']].head(5)
```

## 3. Vocabulary Statistics (Token Length Distribution)

```
import matplotlib.pyplot as plt
import seaborn as sns

processed_df['token_count'] = processed_df['cleaned_text'].apply(lambda x: len(x.split()))

plt.figure(figsize=(8, 4))
sns.histplot(processed_df['token_count'], bins=15, kde=True, color='#00E5FF')
plt.title("Distribution of Token Lengths per Cleaned Headline", fontsize=12, fontweight='bold')
plt.xlabel("Number of Tokens", fontsize=11)
plt.ylabel("Frequency", fontsize=11)
plt.show()

print(f"Mean Token Count: {processed_df['token_count'].mean():.2f}")
print(f"Max Token Count:  {processed_df['token_count'].max()}")
print(f"Min Token Count:  {processed_df['token_count'].min()}")
```

## 4. Preprocessing Conclusion
- All non-informative syntax (URLs, tags, numbers, punctuation) removed.
- Crucial financial terms and negations preserved.
- Output serialized to `data/processed/cleaned_news.csv` ready for TF-IDF feature extraction.
